```python

import pandas as pd

folder = '/content/drive/MyDrive/Mini Project 2/Datasets'

file_path = f'{folder}/IDS_CountryMetaData.csv'

country_data = pd.read_csv(file_path, encoding='cp1252')

print("CSV file imported successfully!")
```

```python
print("Dataset shape:", country_data.shape)

display(country_data.head())
```

```python
# Display the first 5 rows
display(country_data.head())

# Display all column names
print(country_data.columns.tolist())

# Check data types and missing values
country_data.info()
```

```python
# Count duplicate rows
print("Number of duplicate rows:", country_data.duplicated().sum())
```

```python
# Count rows where all 30 columns are null
print("Number of completely empty rows:", country_data.isnull().all(axis=1).sum())
```

```python
print("Unique country codes:", country_data["Code"].nunique())
print("Total rows:", len(country_data))
```

```python
# Check columns where every value is missing
empty_columns = country_data.columns[country_data.isnull().all()]

print("Completely empty columns:", empty_columns.tolist())
```

```python
# Select all numerical columns
numerical_data = country_data.select_dtypes(include='number')

# Display all numerical columns
display(numerical_data)
```
