```python
import pandas as pd

folder = '/content/drive/MyDrive/Mini Project 2/Datasets'
file_path = f'{folder}/IDS_SeriesMetaData.csv'

series_data = pd.read_csv(file_path, encoding='cp1252')

print("Dataset imported successfully!")
print("Dataset shape:", series_data.shape)

display(series_data.head())
```

```python
series_data.info()

print("Column names:", series_data.columns.tolist())
```

```python
# Check duplicate rows
print("Duplicate rows:", series_data.duplicated().sum())

# Check completely empty rows
print("Completely empty rows:",
      series_data.isnull().all(axis=1).sum())

# Check completely empty columns
empty_columns = series_data.columns[series_data.isnull().all()]

print("Completely empty columns:", empty_columns.tolist())
```
