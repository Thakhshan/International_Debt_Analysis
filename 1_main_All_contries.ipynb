  **Now changing the Wide Spread dataframe (debt.data) to long spread dataframe**
  **Because there are many year columns , while using SQL commands we can easily write query for year and debt values**
  **So now changing the 2000 to 2032 year columns to just Year and Debt value columns, so that we can easily write SQL query and can create Power BI Dashboard.**

  **Example for Wide data to Long Data as below picture**


{3C46E3C2-8AB9-45E5-8809-F106F9A8E94A}.png

```python
# Identify year columns
year_columns = [
    col for col in debt_data.columns
    if col.isdigit()
]

# Identify non-year columns
id_columns = [
    col for col in debt_data.columns
    if col not in year_columns
]

# Convert wide format to long format
long_debt_data = debt_data.melt(
    id_vars=id_columns,
    value_vars=year_columns,
    var_name="Year",
    value_name="Debt Value"
)

# Convert Year to integer
long_debt_data["Year"] = long_debt_data["Year"].astype(int)

# Display the result
print(long_debt_data.head())
print("Shape:", long_debt_data.shape)
```

**Now there are many null values in Debt column, as the main purpose of this analysis is to analyse the debt , but if there is no debt so i removed the empty debt column values**

**and it is easy to handle when we write SQL query and Power BI**

```python
# Remove rows where Debt Value is missing
long_debt_cleaned = long_debt_data.dropna(
    subset=["Debt Value"]
).copy()

# Check the results
print("Before:", long_debt_data.shape)
print("After:", long_debt_cleaned.shape)

print(
    "Missing debt values:",
    long_debt_cleaned["Debt Value"].isna().sum()
)
```
